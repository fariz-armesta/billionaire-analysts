```python
mask = df_billionaire["age"].isnull()

birth_year = pd.to_datetime(df_billionaire.loc[mask, "birth_date"], errors="coerce").dt.year
df_billionaire.loc[mask, "age"] = df_billionaire.loc[mask, "year"] - birth_year

df_billionaire["age"] = df_billionaire["age"].fillna(df_billionaire["age"].mean())
df_billionaire["age"] = df_billionaire["age"].astype(int)
df_billionaire = df_billionaire.drop(columns=["birth_date"])
```

```python
df_billionaire["age"] = df_billionaire["age"].fillna(df_billionaire["age"].mean().astype(str).split(".")[0])
df_billionaire["age"] = df_billionaire["age"].astype(int)
df_billionaire["rank"] = df_billionaire["rank"].astype(int)
df_billionaire["organization_name"] = df_billionaire["organization_name"].fillna("Unspecified Organization")
df_billionaire["position_in_organization"] = df_billionaire["position_in_organization"].fillna("Unspecified Position")
df_billionaire["wealth_status"] = df_billionaire["wealth_status"].fillna("Unknown")
df_billionaire["gender"] = df_billionaire["gender"].fillna("Group/Family")
df_billionaire["business_category"] = df_billionaire["business_category"].fillna("Unspecified Category")
df_billionaire["business_industries"] = df_billionaire["business_industries"].fillna("['Unspecified']")
df_billionaire["city_of_residence"] = df_billionaire["city_of_residence"].fillna(df_billionaire["country_of_citizenship"])
df_billionaire["self_made"] = df_billionaire["self_made"].astype("boolean")
```
